```python
import numpy as np
import matplotlib.pyplot as plt
N = 1000

passos = np.random.choice([-1, 1], size=N)

posicao = np.cumsum(passos)

plt.plot(posicao)
plt.xlabel("Número de passos")
plt.ylabel("Posição")
plt.title("Passeio aleatório")
plt.show()
```

```python
N = 1000
M = 100

posicoes = np.zeros(M)

for i in range(N):
    passos = np.random.choice([-1, 1], size=M)
    posicoes += passos

plt.hist(posicoes, bins=20, density=True)
plt.xlabel("Posição")
plt.ylabel("Probabilidade")
plt.title("Difusão por passeio aleatório")
plt.show()
```
